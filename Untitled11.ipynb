```python
# Amino-acid property table
# Charge: standard physiological-pH side-chain charge
# Hydrophobicity: Kyte & Doolittle (1982) scale
# Size: residue volume in cubic angstroms, Zamyatnin (1972)
amino_acid_properties = {
    'A': {'charge': 0, 'hydrophobicity': 1.8, 'size': 88.6},
    'R': {'charge': 1, 'hydrophobicity': -4.5, 'size': 173.4},
    'N': {'charge': 0, 'hydrophobicity': -3.5, 'size': 114.1},
    'D': {'charge': -1, 'hydrophobicity': -3.5, 'size': 111.1},
    'C': {'charge': 0, 'hydrophobicity': 2.5, 'size': 108.5},
    'Q': {'charge': 0, 'hydrophobicity': -3.5, 'size': 143.8},
    'E': {'charge': -1, 'hydrophobicity': -3.5, 'size': 138.4},
    'G': {'charge': 0, 'hydrophobicity': -0.4, 'size': 60.1},
    'H': {'charge': 1, 'hydrophobicity': -3.2, 'size': 153.2},
    'I': {'charge': 0, 'hydrophobicity': 4.5, 'size': 166.7},
    'L': {'charge': 0, 'hydrophobicity': 3.8, 'size': 166.7},
    'K': {'charge': 1, 'hydrophobicity': -3.9, 'size': 168.6},
    'M': {'charge': 0, 'hydrophobicity': 1.9, 'size': 162.9},
    'F': {'charge': 0, 'hydrophobicity': 2.8, 'size': 189.9},
    'P': {'charge': 0, 'hydrophobicity': -1.6, 'size': 112.7},
    'S': {'charge': 0, 'hydrophobicity': -0.8, 'size': 89.0},
    'T': {'charge': 0, 'hydrophobicity': -0.7, 'size': 116.1},
    'W': {'charge': 0, 'hydrophobicity': -0.9, 'size': 227.8},
    'Y': {'charge': 0, 'hydrophobicity': -1.3, 'size': 193.6},
    'V': {'charge': 0, 'hydrophobicity': 4.2, 'size': 140.0},
}
```

```python
# Compare two aligned 495-aa sequences and compute pair-level features
def compute_features(seq1, seq2, properties):
    n_sub = 0
    d_charge = 0
    d_hydro = 0
    d_size = 0
    n_skipped = 0

    for aa1, aa2 in zip(seq1, seq2):
        if aa1 == aa2:
            continue
        if aa1 == '-' or aa2 == '-':
            n_skipped += 1  # alignment gap, not a true substitution
            continue
        if aa1 not in properties or aa2 not in properties:
            n_skipped += 1  # ambiguous residue (X, B, Z...)
            continue
        n_sub += 1
        d_charge += abs(properties[aa1]['charge'] - properties[aa2]['charge'])
        d_hydro += abs(properties[aa1]['hydrophobicity'] - properties[aa2]['hydrophobicity'])
        d_size += abs(properties[aa1]['size'] - properties[aa2]['size'])

    return n_sub, d_charge, d_hydro, d_size, n_skipped
```

```python
# Build the training table
feature_rows = []
for _, row in labels.iterrows():
    serum_seq = sequences[row['serum']]
    virus_seq = sequences[row['virus']]
    n_sub, d_charge, d_hydro, d_size, n_skipped = compute_features(serum_seq, virus_seq, amino_acid_properties)
    feature_rows.append({
        'serum': row['serum'],
        'virus': row['virus'],
        'num_substitutions': n_sub,
        'total_abs_charge_change': d_charge,
        'total_abs_hydrophobicity_change': d_hydro,
        'total_abs_size_change': d_size,
        'skipped_positions': n_skipped,
        'antigenic_distance': row['antigenic_distance'],
    })

training_table = pd.DataFrame(feature_rows)
print(training_table.shape)
training_table.head()
```

# **Step 9 — Random Forest Regressor + GroupKFold Cross-Validation**

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GroupKFold
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
import numpy as np

feature_cols = ['num_substitutions', 'total_abs_charge_change',
                 'total_abs_hydrophobicity_change', 'total_abs_size_change']

X = training_table[feature_cols].values
y = training_table['antigenic_distance'].values

# Group by virus name so no single strain leaks across train/test split
groups = training_table['virus'].values

print("Unique strains used as CV groups:", len(np.unique(groups)))
```

```python
# Grouped 5-fold cross-validation
gkf = GroupKFold(n_splits=5)

rmse_scores = []
r2_scores = []
mae_scores = []

for fold, (train_idx, test_idx) in enumerate(gkf.split(X, y, groups)):
    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]

    model = RandomForestRegressor(n_estimators=500, random_state=42)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    r2 = r2_score(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)

    rmse_scores.append(rmse)
    r2_scores.append(r2)
    mae_scores.append(mae)

    print(f"Fold {fold+1}: RMSE={rmse:.3f}, R2={r2:.3f}, MAE={mae:.3f}")

print("\nMean RMSE:", np.mean(rmse_scores), "+/-", np.std(rmse_scores))
print("Mean R2:  ", np.mean(r2_scores), "+/-", np.std(r2_scores))
print("Mean MAE: ", np.mean(mae_scores), "+/-", np.std(mae_scores))
```

```python
# Train the final model on all data (this is the model used later for Bangladeshi strains)
final_model = RandomForestRegressor(n_estimators=500, random_state=42)
final_model.fit(X, y)

# Feature importance - useful for the manuscript's Results/Discussion
importance_table = pd.DataFrame({
    'feature': feature_cols,
    'importance': final_model.feature_importances_
}).sort_values('importance', ascending=False)

print(importance_table)
```

# **overfitting checked**

```python
from sklearn.dummy import DummyRegressor

# Baseline: always predict the training-fold mean
baseline_rmse, baseline_mae = [], []

for fold, (train_idx, test_idx) in enumerate(gkf.split(X, y, groups)):
    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]

    baseline = DummyRegressor(strategy='mean')
    baseline.fit(X_train, y_train)
    y_pred = baseline.predict(X_test)

    baseline_rmse.append(np.sqrt(mean_squared_error(y_test, y_pred)))
    baseline_mae.append(mean_absolute_error(y_test, y_pred))

print("Baseline (mean predictor) - Mean RMSE:", np.mean(baseline_rmse))
print("Baseline (mean predictor) - Mean MAE: ", np.mean(baseline_mae))
```

```python
# Regularized Random Forest: shallower trees, larger leaf size to reduce overfitting
reg_rmse, reg_r2, reg_mae = [], [], []

for fold, (train_idx, test_idx) in enumerate(gkf.split(X, y, groups)):
    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]

    model = RandomForestRegressor(
        n_estimators=300,
        max_depth=4,
        min_samples_leaf=10,
        random_state=42
    )
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    r2 = r2_score(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)

    reg_rmse.append(rmse)
    reg_r2.append(r2)
    reg_mae.append(mae)

    print(f"Fold {fold+1}: RMSE={rmse:.3f}, R2={r2:.3f}, MAE={mae:.3f}")

print("\nRegularized RF - Mean RMSE:", np.mean(reg_rmse))
print("Regularized RF - Mean R2:  ", np.mean(reg_r2))
print("Regularized RF - Mean MAE: ", np.mean(reg_mae))
```

```python
from sklearn.model_selection import GridSearchCV
from scipy.stats import spearmanr, pearsonr

param_grid = {
    'max_depth': [3, 4, 5, 6],
    'min_samples_leaf': [5, 10, 15, 20],
    'n_estimators': [300]
}

# Custom grouped CV using our existing GroupKFold splits
gkf = GroupKFold(n_splits=5)

grid_search = GridSearchCV(
    estimator=RandomForestRegressor(random_state=42),
    param_grid=param_grid,
    cv=gkf.split(X, y, groups),
    scoring='neg_root_mean_squared_error',
    n_jobs=-1
)
grid_search.fit(X, y)

print("Best params:", grid_search.best_params_)
print("Best CV RMSE:", -grid_search.best_score_)
```

```python
# Re-run grouped CV with the best hyperparameters, now also tracking correlation
best_params = grid_search.best_params_

final_rmse, final_r2, final_mae = [], [], []
final_pearson, final_spearman = [], []

for fold, (train_idx, test_idx) in enumerate(gkf.split(X, y, groups)):
    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]

    model = RandomForestRegressor(random_state=42, **best_params)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    final_rmse.append(np.sqrt(mean_squared_error(y_test, y_pred)))
    final_r2.append(r2_score(y_test, y_pred))
    final_mae.append(mean_absolute_error(y_test, y_pred))
    final_pearson.append(pearsonr(y_test, y_pred)[0])
    final_spearman.append(spearmanr(y_test, y_pred)[0])

    print(f"Fold {fold+1}: RMSE={final_rmse[-1]:.3f}, R2={final_r2[-1]:.3f}, "
          f"Pearson={final_pearson[-1]:.3f}, Spearman={final_spearman[-1]:.3f}")

print("\nMean RMSE:    ", np.mean(final_rmse), "+/-", np.std(final_rmse))
print("Mean R2:      ", np.mean(final_r2), "+/-", np.std(final_r2))
print("Mean MAE:     ", np.mean(final_mae), "+/-", np.std(final_mae))
print("Mean Pearson: ", np.mean(final_pearson), "+/-", np.std(final_pearson))
print("Mean Spearman:", np.mean(final_spearman), "+/-", np.std(final_spearman))
```

```python
def get_envelope_protein(accession):
    handle = Entrez.efetch(db="nucleotide", id=accession, rettype="gb", retmode="text")
    record = SeqIO.read(handle, "genbank")
    handle.close()

    for feature in record.features:
        if feature.type not in ("CDS", "mat_peptide"):
            continue
        product = " ".join(feature.qualifiers.get("product", [""])).lower()
        if "envelope" in product:
            # mat_peptide usually has no direct translation qualifier - extract and translate manually
            translation = feature.qualifiers.get("translation")
            if translation:
                return translation[0]
            seq = feature.extract(record.seq)
            return str(seq.translate(to_stop=True))
    return None
```

```python
from Bio import Align

def extract_e_by_alignment(polyprotein_accession, reference_e_protein):
    handle = Entrez.efetch(db="nucleotide", id=polyprotein_accession, rettype="gb", retmode="text")
    record = SeqIO.read(handle, "genbank")
    handle.close()

    cds_feature = next(f for f in record.features if f.type == "CDS")
    full_translation = cds_feature.qualifiers.get("translation")
    if full_translation:
        polyprotein_seq = full_translation[0]
    else:
        polyprotein_seq = str(cds_feature.extract(record.seq).translate(to_stop=True))

    aligner = Align.PairwiseAligner()
    aligner.mode = "local"
    aligner.open_gap_score = -10
    aligner.extend_gap_score = -0.5
    aligner.match_score = 2
    aligner.mismatch_score = -1

    alignment = aligner.align(polyprotein_seq, reference_e_protein)[0]
    start, end = alignment.aligned[0][0][0], alignment.aligned[0][-1][1]

    return polyprotein_seq[start:end], alignment.score, end - start
```

```python
from Bio.Seq import Seq

def get_six_frame_translations(nt_seq):
    frames = []
    seq = Seq(nt_seq)
    rev_seq = seq.reverse_complement()

    for frame in range(3):
        frames.append(str(seq[frame:].translate(to_stop=False)))
    for frame in range(3):
        frames.append(str(rev_seq[frame:].translate(to_stop=False)))
    return frames
```

```python
def extract_e_protein_bd(nt_seq, reference_e_protein, min_score=400):
    frames = get_six_frame_translations(nt_seq)

    aligner = Align.PairwiseAligner()
    aligner.mode = "local"
    aligner.open_gap_score = -10
    aligner.extend_gap_score = -0.5
    aligner.match_score = 2
    aligner.mismatch_score = -1

    best_score = -1
    best_segment = None

    for frame_seq in frames:
        if len(frame_seq) < 50:
            continue
        alignment = aligner.align(frame_seq, reference_e_protein)[0]
        if alignment.score > best_score:
            start, end = alignment.aligned[0][0][0], alignment.aligned[0][-1][1]
            best_score = alignment.score
            best_segment = frame_seq[start:end]

    if best_score < min_score:
        return None, best_score  # too weak a match - likely doesn't contain full E gene

    return best_segment, best_score
```

```python
# Fix: trim each frame to a multiple of 3 before translation, to avoid the partial-codon warning
def get_six_frame_translations(nt_seq):
    frames = []
    seq = Seq(nt_seq)
    rev_seq = seq.reverse_complement()

    for frame in range(3):
        trimmed = seq[frame:]
        trimmed = trimmed[:len(trimmed) - len(trimmed) % 3]
        frames.append(str(trimmed.translate(to_stop=False)))
    for frame in range(3):
        trimmed = rev_seq[frame:]
        trimmed = trimmed[:len(trimmed) - len(trimmed) % 3]
        frames.append(str(trimmed.translate(to_stop=False)))
    return frames
```

```python
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.svm import SVR
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, WhiteKernel
from sklearn.preprocessing import StandardScaler

models_to_test = {
    "RandomForest (tuned)": RandomForestRegressor(max_depth=3, min_samples_leaf=20, n_estimators=300, random_state=42),
    "Ridge": Ridge(alpha=1.0),
    "GradientBoosting": GradientBoostingRegressor(max_depth=2, min_samples_leaf=20, n_estimators=200, learning_rate=0.05, random_state=42),
    "SVR": SVR(kernel='rbf', C=1.0),
    "GaussianProcess": GaussianProcessRegressor(kernel=RBF() + WhiteKernel(), random_state=42),
}
```

```python
comparison_results = []

for model_name, model in models_to_test.items():
    fold_rmse, fold_r2, fold_mae, fold_spearman = [], [], [], []

    for fold, (train_idx, test_idx) in enumerate(gkf.split(X, y, groups)):
        X_train, X_test = X[train_idx], X[test_idx]
        y_train, y_test = y[train_idx], y[test_idx]

        # SVR and Gaussian Process need scaled features; tree models don't
        if model_name in ["SVR", "GaussianProcess"]:
            scaler = StandardScaler()
            X_train_use = scaler.fit_transform(X_train)
            X_test_use = scaler.transform(X_test)
        else:
            X_train_use, X_test_use = X_train, X_test

        model.fit(X_train_use, y_train)
        y_pred = model.predict(X_test_use)

        fold_rmse.append(np.sqrt(mean_squared_error(y_test, y_pred)))
        fold_r2.append(r2_score(y_test, y_pred))
        fold_mae.append(mean_absolute_error(y_test, y_pred))
        fold_spearman.append(spearmanr(y_test, y_pred)[0])

    comparison_results.append({
        'model': model_name,
        'mean_rmse': np.mean(fold_rmse),
        'mean_r2': np.mean(fold_r2),
        'mean_mae': np.mean(fold_mae),
        'mean_spearman': np.mean(fold_spearman),
        'std_r2': np.std(fold_r2),
    })

comparison_table = pd.DataFrame(comparison_results).sort_values('mean_r2', ascending=False)
print(comparison_table)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 5))
comparison_sorted = comparison_table.sort_values('mean_r2', ascending=True)

ax.barh(comparison_sorted['model'], comparison_sorted['mean_r2'],
        xerr=comparison_sorted['std_r2'], color='steelblue', capsize=4)
ax.set_xlabel('Mean R² (5-fold GroupKFold CV)')
ax.set_title('Model Comparison: Predictive Performance')
ax.axvline(0, color='black', linewidth=0.8)
plt.tight_layout()
plt.savefig('model_comparison.png', dpi=150)
plt.show()
```
